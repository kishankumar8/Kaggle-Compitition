```python
train_data['Arrival Delay in Minutes'].fillna(train_data['Arrival Delay in Minutes'].mean(),inplace=True)
```

```python
train_data.isnull().sum()
```

```python
y=train_data['satisfaction']
x=train_data.drop(['satisfaction'],axis=1)
```

```python
x=pd.get_dummies(x,columns=['Gender','Customer Type','Type of Travel','Class'],drop_first=True)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=DecisionTreeClassifier()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
accuracy=accuracy_score(y_test,y_pred)
print('Accuracy:',accuracy)
```

```python
score=f1_score(y_test,y_pred)
print('F1 Score:',score)
```

```python
cm=confusion_matrix(y_test,y_pred)
print('Confusion Matrix:\n',cm)
```

```python
test_x = pd.get_dummies(test_data, columns=['Gender','Customer Type','Type of Travel','Class'], drop_first=True)
```

```python
test_x = test_x.reindex(columns=x_train.columns, fill_value=0)
```

```python
y_pred=model.predict(test_x)
```

```python
submission=pd.DataFrame({
    'id':test_data['id'],
    'satisfaction':y_pred
})
```
